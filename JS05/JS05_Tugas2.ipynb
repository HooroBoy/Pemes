### 1. Load dan Eksplorasi Data `spam.csv`

```python
import pandas as pd

# Membaca dataset dengan encoding latin-1 karena biasanya file spam.csv memiliki karakter khusus
df = pd.read_csv('/content/spam.csv', encoding='latin-1')

# Membersihkan kolom yang tidak diperlukan jika ada
df = df.dropna(how="any", axis=1)
df.columns = ['label', 'text']

# Menampilkan 5 data teratas
display(df.head())
print(df['label'].value_counts())
```

### 2. Splitting Data & Pembuatan Model Klasifikasi

Kita akan membagi data menjadi training dan testing set, lalu membangun dua model Multinomial Naive Bayes menggunakan `CountVectorizer` dan `TfidfVectorizer` dengan stop words bahasa Inggris.

```python
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn.naive_bayes import MultinomialNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import classification_report, accuracy_score

# Split data (80% train, 20% test)
X_train, X_test, y_train, y_test = train_test_split(df['text'], df['label'], test_size=0.2, random_state=42, stratify=df['label'])

# --- Model 1: CountVectorizer + Multinomial Naive Bayes ---
pipeline_cv = Pipeline([
    ('vectorizer', CountVectorizer(stop_words='english')),
    ('classifier', MultinomialNB())
])

pipeline_cv.fit(X_train, y_train)
y_pred_cv = pipeline_cv.predict(X_test)

# --- Model 2: TF-IDF Vectorizer + Multinomial Naive Bayes ---
pipeline_tfidf = Pipeline([
    ('vectorizer', TfidfVectorizer(stop_words='english')),
    ('classifier', MultinomialNB())
])

pipeline_tfidf.fit(X_train, y_train)
y_pred_tfidf = pipeline_tfidf.predict(X_test)
```

### 3. Evaluasi dan Perbandingan Hasil

Mari kita tampilkan laporan klasifikasi untuk kedua metode untuk membandingkan performanya secara detail.

```python
print("=== EVALUASI MODEL 1: COUNT VECTORIZER ===")
print(f"Accuracy: {accuracy_score(y_test, y_pred_cv):.4f}")
print(classification_report(y_test, y_pred_cv))

print("\n=== EVALUASI MODEL 2: TF-IDF VECTORIZER ===")
print(f"Accuracy: {accuracy_score(y_test, y_pred_tfidf):.4f}")
print(classification_report(y_test, y_pred_tfidf))
```
