```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
df = pd.read_csv('wbc.csv')
```

```python
kolom_dibuang = ['id', 'Unnamed: 32']
df_clean = df.drop(columns=kolom_dibuang)
```

```python
target = df_clean['diagnosis']
fitur_numerik = df_clean.drop(columns=['diagnosis'])
```

```python
label_encoder = LabelEncoder()
target_encoded = label_encoder.fit_transform(target)
df_clean['diagnosis'] = target_encoded
```

```python
scaler = StandardScaler()
fitur_scaled = scaler.fit_transform(fitur_numerik)
```

```python
df_scaled = pd.DataFrame(fitur_scaled, columns=fitur_numerik.columns)
df_scaled.insert(0, 'diagnosis', target_encoded)
```
